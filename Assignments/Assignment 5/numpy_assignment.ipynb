# Array Generation
* ## Generate an array of numbers starting from 5 to 20 with a step of 3 using np.arange.
* ## Create a 3×3 array with random float numbers between 0 and 1 using np.random.rand.
* ## Create a 2×4 array filled with zeros.
* ## Given arr = np.array([[1, 2], [3, 4]]), create a zeros array with the same shape as arr.
* ## Create a 3×3 array filled with ones.
* ## Given arr2 = np.array([[9, 8], [7, 6]]), create an array filled with ones using ones_like.
* ## Create a 2×2 array filled with the number 99.
* ## Use full_like to create an array of shape and type like arr2, filled with the number 5.
* ## Create an array of 10 equally spaced numbers between 0 and 5 using np.linspace.
* ## Create an array with the elements [1, 2, 3.5, 4]. What is the data type of the resulting array? Then convert the array into an integer type and print both arrays and their dtypes.


```python
import numpy as np

arr1 = np.arange(5, 20, 3)
print("1:", arr1)

arr2 = np.random.rand(3, 3)
print("\n2:\n", arr2)

arr3 = np.zeros((2, 4))
print("\n3:\n", arr3)

arr = np.array([[1, 2], [3, 4]])
arr4 = np.zeros_like(arr)
print("\n4:\n", arr4)

arr5 = np.ones((3, 3))
print("\n5:\n", arr5)

arr2_example = np.array([[9, 8], [7, 6]])
arr6 = np.ones_like(arr2_example)
print("\n6:\n", arr6)

arr7 = np.full((2, 2), 99)
print("\n7:\n", arr7)

arr8 = np.full_like(arr2_example, 5)
print("\n8:\n", arr8)

arr9 = np.linspace(0, 5, 10)
print("\n9:\n", arr9)

arr10 = np.array([1, 2, 3.5, 4])

print("\n10:")
print("Original array:", arr10)
print("Original dtype:", arr10.dtype)

arr10_int = arr10.astype(int)

print("Converted array:", arr10_int)
print("Converted dtype:", arr10_int.dtype)
```

# Array Properties
## For arr = np.array([[1, 2, 3], [4, 5, 6]]), print:

* ## Shape

* ## Size

* ## Item size (in bytes)

* ## Total memory (nbytes)


```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

print("Shape:", arr.shape)
print("Size:", arr.size)
print("Item size:", arr.itemsize)
print("Total memory:", arr.nbytes)
```

# Statistics
## Given arr = np.array([10, 30, 20, 40, 50]), calculate:

* ## Minimum value

* ## Maximum value

* ## Mean

* ## Median

* ## Standard Deviation

```python
import numpy as np

arr = np.array([10, 30, 20, 40, 50])

print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
```

# Matrix Operations
## Given matrix = np.array([[1, 2], [3, 4]]), find:

* ## The transpose of the matrix

* ## The determinant (hint: use np.linalg.det())

```python
import numpy as np

matrix = np.array([[1, 2],
                   [3, 4]])

transpose_matrix = matrix.T
print("Transpose:\n", transpose_matrix)

determinant = np.linalg.det(matrix)
print("\nDeterminant:", determinant)
```

# Data Types
* ##  Create an array with values [5.4, 6.3, 7.1] and convert its data type to int.
* ## Print the data type of this array: np.array([True, False, True]).
* ## Create an array with values ["ABBB", "BCC", "CDD"] with data type UNICODE and convert its data type to String.

```python
import numpy as np

arr1 = np.array([5.4, 6.3, 7.1])

print("Original array:", arr1)
print("Original dtype:", arr1.dtype)

arr1_int = arr1.astype(int)

print("Converted array:", arr1_int)
print("Converted dtype:", arr1_int.dtype)

arr2 = np.array([True, False, True])

print("\nBoolean array:", arr2)
print("Boolean dtype:", arr2.dtype)

arr3 = np.array(["ABBB", "BCC", "CDD"], dtype='U')

print("\nUnicode array:", arr3)
print("Unicode dtype:", arr3.dtype)

arr3_str = arr3.astype(str)

print("Converted String array:", arr3_str)
print("Converted dtype:", arr3_str.dtype)
```

## Create a 3×3 matrix with values from 1 to 9 using np.arange and reshape. Then:

* ## Multiply all elements by 2

* ## Find the square root of each element

* ## Find the sum across axis=0 and axis=1

```python
import numpy as np

matrix = np.arange(1, 10).reshape(3, 3)
print("Original Matrix:\n", matrix)

multiplied_matrix = matrix * 2
print("\nMatrix after multiplying by 2:\n", multiplied_matrix)

sqrt_matrix = np.sqrt(matrix)
print("\nSquare root of each element:\n", sqrt_matrix)

sum_axis0 = np.sum(matrix, axis=0)
print("\nSum across axis=0:", sum_axis0)

sum_axis1 = np.sum(matrix, axis=1)
print("Sum across axis=1:", sum_axis1)
```

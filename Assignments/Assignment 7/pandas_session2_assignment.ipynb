### Dataset Description

- market_segment: Market segment designation
- agent, company: Booking agent/company ID (has missing values)
- adr: Average Daily Rate
- reservation_status: Current status

**Known Data Quality Issues:**

- Missing values in: children, country, agent, company

## Part 4: Data Transformation (20 points)
---

### Task 4.1: Replace and Rename (5 points)

1. In 'meal' column, replace 'Undefined' with 'No Meal'
2. Rename columns: 'adr' to 'avg_daily_rate', 'is_canceled' to 'canceled'
3. Verify the changes

```python
df["meal"] = df["meal"].replace("Undefined", "No Meal")

df = df.rename(columns={
    "adr": "avg_daily_rate",
    "is_canceled": "canceled"
})

df[["meal", "avg_daily_rate", "canceled"]].head()
```

### Task 4.2: Create Derived Columns (5 points)

1. Create 'total_nights' = stays_in_weekend_nights + stays_in_week_nights
2. Create 'total_guests' = adults + children + babies
3. Create 'arrival_date' by combining year, month, day columns into a proper date

```python
df["total_nights"] = df["stays_in_weekend_nights"] + df["stays_in_week_nights"]

df["total_guests"] = df["adults"] + df["children"] + df["babies"]

df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str) + "-" +
    df["arrival_date_month"] + "-" +
    df["arrival_date_day_of_month"].astype(str)
)

df[["total_nights", "total_guests", "arrival_date"]].head()
```

### Task 4.3: Binning with cut() and qcut() (5 points)

1. Create 'lead_time_category' using cut():
   - 0-7 days: 'Last Minute'
   - 8-30 days: 'Short Notice'
   - 31-90 days: 'Planned'
   - 90+ days: 'Far Advance'

2. Create 'price_quartile' using qcut() to divide avg_daily_rate into 4 equal groups

```python
df["lead_time_category"] = pd.cut(
    df["lead_time"],
    bins=[0, 7, 30, 90, float("inf")],
    labels=["Last Minute", "Short Notice", "Planned", "Far Advance"],
    include_lowest=True
)

df["price_quartile"] = pd.qcut(
    df["avg_daily_rate"],
    q=4,
    labels=["Low", "Medium", "High", "Premium"]
)

df[["lead_time", "lead_time_category", "avg_daily_rate", "price_quartile"]].head()
```

### Task 4.4: Outlier Detection (5 points)

The avg_daily_rate column may have outliers.

1. Calculate the IQR (Interquartile Range)
2. Find outliers using: lower < Q1 - 1.5*IQR or upper > Q3 + 1.5*IQR
3. How many outliers exist?
4. Cap outliers at the upper/lower bounds (do not remove them)

```python
Q1 = df["avg_daily_rate"].quantile(0.25)
Q3 = df["avg_daily_rate"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["avg_daily_rate"] < lower_bound) |
    (df["avg_daily_rate"] > upper_bound)
]

len(outliers)

df["avg_daily_rate"] = df["avg_daily_rate"].clip(
    lower_bound,
    upper_bound
)
```

## Part 5: String Manipulation (10 points)
---

### Task 5.1: String Operations (5 points)

1. Convert 'arrival_date_month' to lowercase
2. Create 'country_upper' with country codes in uppercase
3. Check if 'market_segment' contains 'Online' (create boolean column 'is_online')

```python
df["arrival_date_month"] = df["arrival_date_month"].str.lower()

df["country_upper"] = df["country"].str.upper()

df["is_online"] = df["market_segment"].str.contains("Online", case=False)

df[["arrival_date_month", "country_upper", "market_segment", "is_online"]].head()
```

### Task 5.2: String Extraction (5 points)

1. Extract the first 2 characters of 'country' as 'country_prefix'
2. Get the length of 'reservation_status' as 'status_length'
3. Replace any spaces in 'distribution_channel' with underscores

```python
df["country_prefix"] = df["country"].str[:2]

df["status_length"] = df["reservation_status"].str.len()

df["distribution_channel"] = df["distribution_channel"].str.replace(" ", "_")

df[["country_prefix", "status_length", "distribution_channel"]].head()
```

## Part 6: One-Hot Encoding (10 points)
---

### Task 6.1: get_dummies() Basics (5 points)

1. Create one-hot encoding for 'hotel' column
2. Create one-hot encoding for 'meal' column with prefix 'meal'
3. How many new columns were created for meal?

```python
hotel_encoded = pd.get_dummies(df["hotel"])

meal_encoded = pd.get_dummies(df["meal"], prefix="meal")

meal_encoded.shape[1]

hotel_encoded.head()
meal_encoded.head()
```

### Task 6.2: Advanced Encoding (5 points)

1. Create encoding for 'market_segment' using drop_first=True (avoid dummy variable trap)
2. Merge these dummy columns back to the main dataframe
3. Verify the new columns exist

```python
market_encoded = pd.get_dummies(
    df["market_segment"],
    prefix="market",
    drop_first=True
)

df = pd.concat([df, market_encoded], axis=1)

market_encoded.head()

df.columns
```

## Part 7: Merging and Reshaping (10 points)
---

### Task 7.1: Create and Merge Summary Tables (5 points)

1. Create a summary table with average avg_daily_rate per hotel type
2. Create a summary table with count of bookings per market_segment
3. Demonstrate merging: Create a small lookup table and merge it with main data

```python
avg_rate_by_hotel = df.groupby("hotel")["avg_daily_rate"].mean().reset_index()

bookings_by_market = df.groupby("market_segment").size().reset_index(name="booking_count")

hotel_ratings = pd.DataFrame({
    "hotel": ["Resort Hotel", "City Hotel"],
    "star_rating": [4, 3]
})

df = df.merge(hotel_ratings, on="hotel", how="left")

avg_rate_by_hotel

bookings_by_market

df[["hotel", "star_rating"]].head()
```

### Task 7.2: Reshaping with Pivot (5 points)

1. Create a pivot table showing average avg_daily_rate by:
   - Rows: hotel type
   - Columns: customer_type

2. Create a pivot showing count of bookings by:
   - Rows: arrival_date_month
   - Columns: hotel

```python
avg_rate_pivot = df.pivot_table(
    values="avg_daily_rate",
    index="hotel",
    columns="customer_type",
    aggfunc="mean"
)

booking_count_pivot = df.pivot_table(
    index="arrival_date_month",
    columns="hotel",
    values="lead_time",
    aggfunc="count"
)

avg_rate_pivot

booking_count_pivot
```
